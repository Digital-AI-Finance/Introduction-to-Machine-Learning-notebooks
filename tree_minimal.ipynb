# A decision tree in three lines

Two measurements, four hundred points, one picture. Run each cell with
**Shift and Enter**.

```python
from sklearn.datasets import make_moons
from sklearn.tree import DecisionTreeClassifier
from sklearn.inspection import DecisionBoundaryDisplay

X, y = make_moons(400, noise=0.3, random_state=0)
tree = DecisionTreeClassifier(max_depth=3).fit(X, y)
DecisionBoundaryDisplay.from_estimator(tree, X, alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

One line makes the data: 400 points, two measurements each, in two arcs that
overlap. One line grows a tree three questions deep. One line colours every
point of the plane by the answer the tree gives there.

Every edge in the picture is horizontal or vertical, because every question a
tree asks is about one measurement on its own. A tree carves the plane into
rectangles and gives one answer inside each. Where a rectangle holds points of
both colours the shade is pale, and it is the share of that rectangle that is
red.

## The same tree, as questions

```python
from sklearn.tree import plot_tree

plot_tree(tree, impurity=False, fontsize=6)
```

`x[1] <= 0.092` is the first question, and the left branch is taken when the
answer is yes. `samples` is how many of the 400 points reached that box, and
`value` counts them by colour, blue first. The answer a box gives is whichever
count is larger.

The seven boxes with no arrow leaving them are the seven rectangles of the
picture above. `value = [0, 53]` is one of them: 53 points, every one of them
red, which is the rectangle drawn at full strength. `value = [42, 15]` is the
palest, 42 blue against 15 red, and it is the band running down the middle.

## One question only

```python
tree = DecisionTreeClassifier(max_depth=1).fit(X, y)
DecisionBoundaryDisplay.from_estimator(tree, X, alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

One question, one straight cut, two rectangles. The upper holds 147 of one
colour against 21 of the other and the lower 179 against 53, so neither is at
full strength and neither is close to pale. `max_depth` is the one dial that
decides how much shape a tree may have.

Change `max_depth=1` to `max_depth=2`, then `5`, then `10`.

## No limit at all

```python
tree = DecisionTreeClassifier().fit(X, y)
DecisionBoundaryDisplay.from_estimator(tree, X, alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

Left alone, the tree keeps asking until every point it was given is on the
correct side. It reaches 47 rectangles and 12 questions deep, nothing is pale
any more, and thin slivers run to the top and bottom of the picture where one
point was fenced off on its own.

Those slivers are the noise in the 400 points, drawn as if it were shape. A
fresh 400 points from the same two arcs would put the noise somewhere else, and
the slivers would be in the wrong place. Depth past three buys detail that
belongs to the points the tree was handed.

## What you can say now

- A decision tree is a chain of questions, each about one measurement, so every
  boundary it draws is at a right angle.
- Each rectangle gives one answer. Pale means the rectangle holds both colours,
  and the shade is the share.
- `max_depth` is the dial. Too little and the tree has no shape; left off and
  it follows the noise.
