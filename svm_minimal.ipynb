# A support vector machine in three lines

Two measurements, four hundred points, one picture. Run each cell with
**Shift and Enter**.

```python
from sklearn.datasets import make_moons
from sklearn.svm import SVC
from sklearn.inspection import DecisionBoundaryDisplay

X, y = make_moons(400, noise=0.3, random_state=0)
machine = SVC(kernel="linear").fit(X, y)
DecisionBoundaryDisplay.from_estimator(machine, X, response_method="predict", alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

One line makes the data: 400 points, two measurements each, in two arcs that
overlap. One line finds the straight line with the widest empty street between
the two colours. One line colours every point of the plane by the side of that
line it falls on.

The boundary is straight, because `kernel="linear"` has one line to give. It
puts 344 of the 400 points on the correct side, and the 56 it puts on the
wrong side are where the two arcs reach around each other.

## The street, and the points that hold it

```python
show = DecisionBoundaryDisplay.from_estimator(machine, X, plot_method="contour", levels=[-1, 0, 1], colors="k", linestyles=["--", "-", "--"])
show.ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
show.ax_.scatter(*machine.support_vectors_.T, s=70, facecolors="none", edgecolors="k")
```

The solid line is the boundary and the two dashed lines are the kerbs of the
street around it. The street is 0.860 wide, and no straight line with a wider
one separates these two colours as well.

The 146 ringed points stand on a kerb, inside the street, or over the line.
Those are the points the machine keeps, and the line rests on them alone: the
other 254 could be moved anywhere outside the street and the picture would not
change.

## A curved boundary

```python
machine = SVC().fit(X, y)
DecisionBoundaryDisplay.from_estimator(machine, X, response_method="predict", alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

Left at its default the machine measures closeness with a bell shaped weight
around each point, `kernel="rbf"`, and the boundary it draws bends. It follows
the two arcs, puts 365 of the 400 on the correct side, and keeps 121 points.

Away from the points the bell weights fade to nothing and every edge of the
picture takes one colour, which is the answer the machine gives where it has
been shown nothing at all.

## How narrow the bell is

```python
machine = SVC(gamma=50).fit(X, y)
DecisionBoundaryDisplay.from_estimator(machine, X, response_method="predict", alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

`gamma` is how narrow that bell is. At 50 each point speaks for a small circle
around itself alone, and the edge of the blue region tightens onto the blue
points, bending in and out around single dots. 298 of the 400 are kept, and it
puts 371 of them on the correct side, six more than the default.

Try `gamma=0.1`, which widens the bell until the boundary is nearly straight
again and 346 land on the correct side.

## A fresh four hundred points

```python
Xnew, ynew = make_moons(400, noise=0.3, random_state=1)

print("default: %.0f of 400" % (SVC().fit(X, y).score(Xnew, ynew) * 400))
print("gamma=50: %.0f of 400" % (SVC(gamma=50).fit(X, y).score(Xnew, ynew) * 400))
```

A fresh 400 points from the same two arcs settle it. The default gets 360 of
them right and `gamma=50` gets 353, having beaten it by six on the points it
was handed. Those bends are the noise in the first 400 points, drawn as if it
were shape.

Try `C=0.01` in either cell. A crossing becomes so cheap that all 400 points
are kept, the boundary flattens, and 340 of the 400 land on the correct side.

## What you can say now

- A support vector machine takes the boundary with the widest empty street
  around it, and the points on a kerb or inside the street decide where it
  runs.
- `kernel="linear"` gives one straight line. The default `rbf` bends the
  boundary around the points it was given.
- `gamma` is how far one point's influence reaches and `C` is the price of a
  point inside the street. Narrow the bell and the boundary follows the noise,
  bending around single points.
