# A random forest in three lines

Two measurements, four hundred points, one picture. Run each cell with
**Shift and Enter**.

```python
from sklearn.datasets import make_moons
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import DecisionBoundaryDisplay

X, y = make_moons(400, noise=0.3, random_state=0)
forest = RandomForestClassifier(n_estimators=100).fit(X, y)
DecisionBoundaryDisplay.from_estimator(forest, X, alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

One line makes the data: 400 points, two measurements each, in two arcs that
overlap. One line grows a hundred trees on it. One line colours every point of
the plane by what the hundred trees say about it.

Deep red is where all hundred vote red, deep blue where all hundred vote blue,
and the pale band between them is where they split. The width of that band is
where the trees disagreed.

## Fewer trees

```python
forest = RandomForestClassifier(n_estimators=1).fit(X, y)
DecisionBoundaryDisplay.from_estimator(forest, X, alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

One tree has nothing to be pale about. The map goes to two flat colours with a
hard edge, and the edge is a staircase, because every question a tree asks is
about one measurement and cuts at a right angle.

Run the cell again without changing anything. The staircase moves, because the
tree is grown on a fresh draw of the rows each time. Run the hundred tree cell
twice and its picture barely moves: that steadiness is what the other
ninety-nine trees buy.

Try `n_estimators=3` and `n_estimators=10` in the same cell.

## Shallower trees

```python
forest = RandomForestClassifier(n_estimators=100, max_depth=2).fit(X, y)
DecisionBoundaryDisplay.from_estimator(forest, X, alpha=0.4, cmap="coolwarm").ax_.scatter(*X.T, c=y, cmap="coolwarm", s=14)
```

`max_depth` is how many questions a tree may ask before it answers. At 2 the
boundary is a few straight steps and the slivers are gone. At 1 it is a single
straight cut. Left alone, a tree asks as many questions as it needs to put
every point it was given on the correct side, which is where the slivers came
from.

Change `max_depth=2` to `max_depth=1`, then to `max_depth=5`, then delete the
argument.

## What you can say now

- A random forest grows many trees, each on its own resample of the rows, and
  answers by majority.
- The colour is the share of the vote. A pale band means the trees split, and
  its width is the model's own account of where it is unsure.
- `n_estimators` buys steadiness. `max_depth` buys detail, and past a point the
  detail is the noise in the rows the trees were given.
