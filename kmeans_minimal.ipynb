# K-means in three lines

Two measurements, four hundred points, one picture. Run each cell with
**Shift and Enter**.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

X, _ = make_blobs(n_samples=400, centers=4, random_state=0)
groups = KMeans(n_clusters=4, n_init=10, random_state=0).fit(X)
plt.scatter(*X.T, c=groups.labels_, cmap="coolwarm", s=14)
```

One line makes the data: 400 points, two measurements each, in four lumps. One
line finds four centres. One line colours every point by the centre it came
out nearest.

Nothing told the method where the lumps are. It was given the number four and
the points, and the colours are what it settled on.

## The four centres

```python
plt.scatter(*X.T, c=groups.labels_, cmap="coolwarm", s=14)
plt.scatter(*groups.cluster_centers_.T, c="black", marker="x", s=90)

print("points in each group:", [int((groups.labels_ == k).sum()) for k in range(4)])
print("total squared distance to the nearest centre: %.1f" % groups.inertia_)
```

Each cross is the average of the points that took it, and each point takes the
cross it is nearest. Those two sentences are the whole method: start the
crosses somewhere, give every point its nearest, move each cross to the average
of what it got, and go again until nothing moves.

The four groups hold 99, 107, 92 and 102 points, and the total squared distance
from the points to their own centre is 702.6. That number is what the method
makes small.

## The number you choose

```python
groups = KMeans(n_clusters=2, n_init=10, random_state=0).fit(X)
plt.scatter(*X.T, c=groups.labels_, cmap="coolwarm", s=14)
plt.scatter(*groups.cluster_centers_.T, c="black", marker="x", s=90)
```

Two centres divide the same 400 points 138 against 262, and the total squared
distance rises to 2005.8. One lump goes whole to each centre and the boundary
runs through the other two, splitting one of them 31 against 69. The method
takes any number you give it and never says that the number is wrong: at 7 it
cuts the same four lumps into seven, and it will go to 400.

Change `n_clusters=2` to `n_clusters=3`, then `7`.

## Where the fall stops paying

```python
for k in range(1, 8):
    print(k, round(KMeans(n_clusters=k, n_init=10, random_state=0).fit(X).inertia_, 1))
```

Each extra centre lowers the total: 4219.3 with one, 2005.8 with two, 1174.7
with three, 702.6 with four. Then the falls run 88.6, 77.7 and 67.5, an order
of magnitude under the 472.1 that the fourth centre bought.

The bend at four is the whole of the evidence for four, and four is the number
these points were made with.

## The two arcs

```python
from sklearn.datasets import make_moons

Xm, ym = make_moons(400, noise=0.3, random_state=0)
groups = KMeans(n_clusters=2, n_init=10, random_state=0).fit(Xm)
plt.scatter(*Xm.T, c=groups.labels_, cmap="coolwarm", s=14)
plt.scatter(*groups.cluster_centers_.T, c="black", marker="x", s=90)
```

These are the 400 points the decision tree and the random forest notebooks
work on, and a person sees the two arcs at once. K-means gives every point the
nearest of two centres, so the line between its groups is straight and it runs
across both arcs: one group takes 162 points of the upper arc and 59 of the
lower, the other takes 141 of the lower and 38 of the upper.

A method that only draws round groups finds round groups. The two arcs are
there, and this rule has no way to say them.

## What you can say now

- K-means gives every point the nearest of `n_clusters` centres and moves each
  centre to the average of what it got, until nothing moves.
- `inertia_` is the total squared distance to the nearest centre, and it falls
  with every centre added. Where the fall flattens is the evidence for a
  number: 472.1 for the fourth centre here, 88.6 for the fifth.
- The groups are round by construction. Two arcs get cut across.
